```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
import gensim.downloader as api

model = api.load("word2vec-google-news-300")

model # es un diccionario
```

```python
vec = lambda word: model[word]

vec("word")
```

```python
vec("dog")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

cosine_similarity([vec("puppy")], [vec("wolf")]) # comparacion de similitud con esas palabras
```

```python
vec_man = vec("man")
vec_woman = vec("woman")
dist = vec_woman - vec_man
```

```python
vec_king = vec("king")
x = vec_king + dist
```

```python
cosine_similarity([x], [vec("queen")])
```

```python
model.most_similar(x) # entre todos los modelos cual es el mas similar
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components = 2)
```

```python
vman = pca.fit_transform([vec_man])
vwoman = pca.fit_transform([vec_woman])
vking = pca.fit_transform([vec_king])
vqueen = pca.fit_transform([vec("queen")])
vx = pca.fit_transform([x])
```
