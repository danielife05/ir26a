### Paso 1: Calcular la matriz TF-IDF para el corpus `data/01_corpus_turismo_500.txt`

```python
# Importar librerías
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

documentos = []
path = "C:/Users/DELL/OneDrive - Escuela Politécnica Nacional/DANIEL/EPN/SÉPTIMO SEMESTRE/RI/CODIGO/ir26a-danielFlores/data/"

# Leer corpus
with open(path + '01_corpus_turismo_500.txt', 'r', encoding='utf-8') as f:
    documentos = f.readlines()

# Crear vectorizador TF-IDF
vectorizer = TfidfVectorizer()

# Calcular matriz TF-IDF
matriz_tfidf = vectorizer.fit_transform(documentos)

# Ver dimensiones
print(f"Documentos: {matriz_tfidf.shape[0]}")
print(f"Términos únicos: {matriz_tfidf.shape[1]}")

# Convertir a DataFrame para visualizar
df_tfidf = pd.DataFrame(
    matriz_tfidf.toarray(),
    columns=vectorizer.get_feature_names_out()
)

# Ver primeras filas
print(df_tfidf.head())
```

### Paso 2: Construir el corpus `Gutenberg 1000`

El corpus `Gutenberg 1000` es un corpus compuesto por 1000 libros de Gutenberg Project

```python
import os

# Ruta donde están los libros
path_1k = "C:/Users/DELL/OneDrive - Escuela Politécnica Nacional/DANIEL/EPN/SÉPTIMO SEMESTRE/RI/CODIGO/ir26a-danielFlores/data/normalized/"

# Leer todos los archivos
documentos_gutenberg = []

for archivo in os.listdir(path_1k):
    ruta_completa = os.path.join(path_1k, archivo)
    
    with open(ruta_completa, 'r', encoding='utf-8') as f:
        contenido = f.read()
        documentos_gutenberg.append(contenido)

# Verificar
print(f"Total libros cargados: {len(documentos_gutenberg)}")
print(f"Primeros 200 caracteres del libro 1:\n{documentos_gutenberg[0][:200]}")
```

### Paso 3: Calcular la matriz TF-IDF para el corpus `Gutenberg 1000`

```python
# Crear vectorizador TF-IDF
vectorizer_gutenberg = TfidfVectorizer(
    min_df=2,
    max_df=0.8
)

# Calcular matriz TF-IDF
matriz_tfidf_gutenberg = vectorizer_gutenberg.fit_transform(documentos_gutenberg)

# Ver dimensiones
print(f"Documentos: {matriz_tfidf_gutenberg.shape[0]}")
print(f"Términos únicos: {matriz_tfidf_gutenberg.shape[1]}")
print(f"Sparsidad: {(1 - matriz_tfidf_gutenberg.nnz / (matriz_tfidf_gutenberg.shape[0] * matriz_tfidf_gutenberg.shape[1])) * 100:.2f}%")

# Vocabulario aprendido
vocabulario = vectorizer_gutenberg.get_feature_names_out()
print(f"\nPrimeras 20 palabras del vocabulario:\n{vocabulario[:20]}")
```

### Paso 4: Programar una función `buscar()` para el corpus `Gutenberg 1000`

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

query = "love and romance in england"
def buscar(query, top_n=10):
    # Transformar query a vector TF-IDF
    query_vector = vectorizer_gutenberg.transform([query])
    
    # Calcular similitud coseno
    similitudes = cosine_similarity(query_vector, matriz_tfidf_gutenberg)
    
    # Obtener índices ordenados
    indices_ordenados = similitudes[0].argsort()[::-1]
    
    # Top N resultados
    top_indices = indices_ordenados[:top_n]
    top_scores = similitudes[0][top_indices]
    
    # Mostrar resultados
    print(f"Query: '{query}'\n")
    for i, (idx, score) in enumerate(zip(top_indices, top_scores), 1):
        print(f"{i}. Libro {idx} | Score: {score:.4f}")
        print(f"   Preview: {documentos_gutenberg[idx][:150]}...\n")
    
    return top_indices, top_scores

# Probar la función
buscar(query, top_n=5)
```
