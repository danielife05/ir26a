### Paso 1: Calcular la matriz TF-IDF para el corpus `data/01_corpus_turismo_500.txt`

```python
# ============================================================
# PASO 1: Matriz TF-IDF para corpus de turismo (500 documentos)
#
# TF-IDF = TF * IDF
#   TF  (Term Frequency):  cuántas veces aparece el término en el doc
#   IDF (Inverse Doc Freq): qué tan raro es el término en todo el corpus
#   → Resultado: palabras frecuentes en doc pero raras en corpus = peso alto
# ============================================================
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

path = "C:/Users/DELL/OneDrive - Escuela Politécnica Nacional/DANIEL/EPN/SÉPTIMO SEMESTRE/RI/CODIGO/ir26a-danielFlores/data/"

# Cada línea del archivo = un documento
with open(path + '01_corpus_turismo_500.txt', 'r', encoding='utf-8') as f:
    documentos = f.readlines()

# TfidfVectorizer: tokeniza, calcula TF*IDF y devuelve matriz dispersa
# fit_transform = aprender vocabulario + transformar documentos a vectores
vectorizer = TfidfVectorizer()
matriz_tfidf = vectorizer.fit_transform(documentos)

# matriz_tfidf.shape → (filas=documentos, columnas=términos únicos)
print(f"Documentos: {matriz_tfidf.shape[0]}")
print(f"Términos únicos: {matriz_tfidf.shape[1]}")

# toarray(): convierte matriz dispersa → matriz densa (numpy)
# get_feature_names_out(): lista de términos del vocabulario (columnas)
df_tfidf = pd.DataFrame(
    matriz_tfidf.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print(df_tfidf.head())
```

### Paso 2: Construir el corpus `Gutenberg 1000`

El corpus `Gutenberg 1000` es un corpus compuesto por 1000 libros de Gutenberg Project

```python
# ============================================================
# PASO 2: Construir corpus Gutenberg 1000
#
# Cada archivo .txt = un libro = un documento del corpus
# Se carga el contenido completo (no línea a línea como turismo)
# ============================================================
import os
# Ruta donde están los libros
path_1k = "C:/Users/DELL/OneDrive - Escuela Politécnica Nacional/DANIEL/EPN/SÉPTIMO SEMESTRE/RI/CODIGO/ir26a-danielFlores/data/normalized/"

# Leer todos los archivos
documentos_gutenberg = []

# os.listdir: lista todos los archivos en el directorio
for archivo in os.listdir(path_1k):
    ruta_completa = os.path.join(path_1k, archivo)
    #  os.path.join: une ruta + nombre → "data/normalized/libro1.txt"
    
    with open(ruta_completa, 'r', encoding='utf-8') as f:
        contenido = f.read() # # lee libro completo como string
        documentos_gutenberg.append(contenido)

# Verificar
print(f"Total libros cargados: {len(documentos_gutenberg)}")
print(f"Primeros 200 caracteres del libro 1:\n{documentos_gutenberg[0][:200]}")
```

### Paso 3: Calcular la matriz TF-IDF para el corpus `Gutenberg 1000`

```python
# ============================================================
# PASO 3: Matriz TF-IDF para Gutenberg 1000
#
# Parámetros clave del vectorizador:
#   max_features=5000 → solo las 5000 palabras más frecuentes
#                        (evita matrices enormes, similar a PCA en ML)
#   min_df=2          → ignora palabras que aparecen en < 2 libros
#                        (elimina typos y términos irrelevantes)
#   max_df=0.8        → ignora palabras en > 80% de libros
#                        (equivale a stop words automáticas)
#   stop_words='english' → elimina "the", "and", "is"...
#   token_pattern     → solo acepta palabras con letras [a-z]
#                        (descarta números y símbolos)
# ============================================================
print("Procesando 1000 libros...")

vectorizer_gutenberg = TfidfVectorizer(
    # max_features=5000,
    min_df=2,
    max_df=0.8,
    stop_words='english',
)

# fit_transform sobre 1000 libros → matriz (1000 x 5000)
matriz_tfidf_gutenberg = vectorizer_gutenberg.fit_transform(documentos_gutenberg)

# Sparsidad: % de ceros en la matriz
# Una matriz TF-IDF siempre es dispersa: cada libro usa solo una fracción del vocabulario
# En ML: matrices dispersas se usan igual en Naive Bayes, SVM de texto
print("Listo!")
print(f"Documentos: {matriz_tfidf_gutenberg.shape[0]}")
print(f"Términos únicos: {matriz_tfidf_gutenberg.shape[1]}")
print(f"Sparsidad: {(1 - matriz_tfidf_gutenberg.nnz / (matriz_tfidf_gutenberg.shape[0] * matriz_tfidf_gutenberg.shape[1])) * 100:.2f}%")

vocabulario = vectorizer_gutenberg.get_feature_names_out()
print(f"\nPrimeras 20 palabras del vocabulario:\n{vocabulario[:20]}")
```

### Paso 4: Programar una función `buscar()` para el corpus `Gutenberg 1000`

```python
# ============================================================
# PASO 4: Motor de búsqueda con similitud coseno
#
# Flujo:
#   1. query (texto) → lematizar → vector TF-IDF
#   2. calcular cos(θ) entre query y cada uno de los 1000 libros
#   3. ordenar por score descendente → devolver top N
#
# Similitud coseno (Álgebra Lineal):
#   cos(θ) = (A · B) / (||A|| * ||B||)
#   → 1.0 = documentos idénticos
#   → 0.0 = sin palabras en común
# ============================================================

from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
import nltk

nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)
nltk.download('punkt', quiet=True)

lemmatizer = WordNetLemmatizer()

def preprocess_lematizado(text):
    tokens = word_tokenize(text.lower())
    # word_tokenize: ["love", "and", "romances"] 
    
    lemmas = [lemmatizer.lemmatize(t) for t in tokens if t.isalpha()]
    # lemmatize: "romances"→"romance", "running"→"run"
    # t.isalpha(): filtra "'s", ",", "." → consistente con token_pattern del vectorizer
    
    return " ".join(lemmas)

def buscar(query, top_n=10):
    query_procesada = preprocess_lematizado(query)
    
    # transform (NO fit_transform): usa vocabulario ya aprendido del corpus
    # si se usara fit_transform → crearía vocabulario nuevo solo con la query → error
    query_vector = vectorizer_gutenberg.transform([query_procesada])
    
    # similitudes: array de shape (1, 1000) → un score por libro
    similitudes = cosine_similarity(query_vector, matriz_tfidf_gutenberg)
    
    # argsort: devuelve índices que ORDENARÍAN el array de menor a mayor
    # [::-1]: invierte → mayor a menor
    indices_ordenados = similitudes[0].argsort()[::-1]
    
    top_indices = indices_ordenados[:top_n]   # primeros N índices
    top_scores  = similitudes[0][top_indices] # scores de esos N libros
    
    print(f"Query original:   '{query}'")
    print(f"Query lematizada: '{query_procesada}'\n")
    
    for i, (idx, score) in enumerate(zip(top_indices, top_scores), 1):
        # zip: combina [(idx1, score1), (idx2, score2), ...]
        # enumerate(..., 1): contador desde 1
        print(f"{i}. Libro {idx} | Score: {score:.4f}")
        print(f"   Preview: {documentos_gutenberg[idx][:150]}...\n")
    
    return top_indices, top_scores

buscar("love and romance in england", top_n=5)
```
