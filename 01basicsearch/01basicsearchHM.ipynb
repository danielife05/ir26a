### Carga en memoria
Iteramos sobre los 100 archivos y guardamos todo su contenido en un diccionario en memoria para un fácil acceso en futuras cajas de código.

```python
import os
path = r"C:\Users\DELL\OneDrive - Escuela Politécnica Nacional\DANIEL\EPN\SÉPTIMO SEMESTRE\RI\CODIGO\ir26a\data\gutenberg\new_data"
corpus = os.listdir(path)

# Diccionario para almacenar en memoria el contenido de cada documento
textos = {}
# Recorrer a través de cada nombre de archivo disponible en la carpeta
for nombre_archivo in corpus:
    # Construir la ruta completa al archivo iterado actualmente
    file_path = os.path.join(path, nombre_archivo)
    # Abrir archivo y leer contenido
    with open(file_path, "r", encoding="utf-8", errors="ignore") as file:
        # Guardar el contenido en el diccionario con la llave siendo su nombre
        textos[nombre_archivo] = file.read()
        # Imprimir un divisor visual e imprimir una previsualización de 500 caracteres del texto
        print(f"--- CONTENIDO DE {nombre_archivo} ---")
        print(textos[nombre_archivo][:500] + "\n")
```

### Búsqueda temporal cronometrada de una palabra
Buscamos la palabra 'Daniel' en todo el corpus previamente cargado y medimos el tiempo de ejecución.

```python
# Importar la libería time para cronometrar nuestras búsquedas
import time
# Definir el término que queremos buscar
query = "Daniel"
# Iniciar cronómetro de búsqueda
start_time = time.time()
# Lista para acumular los nombres de los archivos donde el término es encontrado
resultados = []

# Iterar cruzando tanto el nombre como el contenido iterando cada ítem en el diccionario
for nombre_archivo, contenido_texto in textos.items():
    # Verificar si el término exacto forma parte del string completo
    if query in contenido_texto:
        print(f"El archivo {nombre_archivo} SÍ contiene la palabra '{query}'")
        resultados.append(nombre_archivo)
    else:
        print(f"El archivo {nombre_archivo} NO contiene la palabra '{query}'")

# Detener cronómetro de búsqueda 
end_time = time.time()
# Mostrar al usuario cuánto tiempo se tardó en segundos y los documentos identificados
print(f"\nTiempo de búsqueda: {end_time - start_time:.4f} segundos")
print(f"Archivos encontrados: {resultados}")
```

### Encapsular lógica en Función
Definimos una función estandarizada para buscar un término puntual dentro de nuestro directorio.

```python
# Función que encapsula nuestra lógica de búsqueda de texto básico
def buscar(query, contenidos):
    resultados = []
    # Iterar por el diccionario de contenidos para comprobar las coincidencias
    for key in contenidos.keys():
        if query in contenidos[key]:
            resultados.append(key)
    return resultados
```

### Extracción de Palabras (Tokens) en Archivo Simple
Separamos el texto de `84.txt` por espacios en blanco y contamos el número total de palabras generadas.

```python
# Abrir puntualmente el archivo 84.txt para contabilizar las palabras reales que posee
with open(path + "\\84.txt", encoding="utf-8", errors="ignore") as f:
    contenido = f.read()
    # Separar el texto crudo según los espacios, lo cual genera una lista de los tokens/palabras
    palabras = contenido.split()

# Imprimir largo (tamaño) total de la lista de palabras generadas
print(len(palabras))
```

### Conteo Global en todo el Corpus
Script añadido para leer uno a uno cada archivo dentro del corpus mediante iteración recursiva y calcular las palabras totales que contienen.

```python
# Iniciar contador de palabras totales en 0
total_palabras = 0
# Recorrer cada archivo en el corpus
for nombre_archivo in corpus:
    # Construir ruta y leer el archivo
    ruta_archivo = os.path.join(path, nombre_archivo)
    with open(ruta_archivo, "r", encoding="utf-8", errors="ignore") as file:
        contenido = file.read()
        # Sumar la cantidad de palabras del documento actual
        total_palabras += len(contenido.split())

# Imprimir el tamaño total de palabras a través de todo el corpus
print(f"Total de palabras en los {len(corpus)} archivos: {total_palabras}")
```

### Búsqueda Masiva de Rendimiento
Tomamos cada palabra encontrada previamente y la buscamos unitariamente en los 100 archivos del corpus para evaluar el costo computacional expresado en milisegundos.

```python
# Medir tiempo de otra tarea más demandante
start_time = time.time()
# Iterar por cada palabra dentro de nuestro arreglo generado en el bloque anterior
for p in palabras:
    # Buscar esa única palabra a lo largo de TODOS los 100 textos que ya están cargados
    buscar(p, textos) # Reemplazado 'contenidos' por 'textos', que es donde guardaste los archivos
end_time = time.time()

# Imprimir el largo de tiempo acumulado en las búsquedas iteradas en términos de milisegundos
print(f"tiempo de búsqueda: {(end_time - start_time) * 1000:.2f} milisegundos")
```

### Extracción de Palabras Únicas (Vocabulario)
Reutilizamos el texto pero implementamos un conjunto (`set`) para purgar elementos duplicados y conocer exclusivamente las palabras únicas de vocabulario.

```python
# Volver a abrir el mismo archivo ahora para obtener palabras únicas
with open(path + "\\84.txt", "r", encoding="utf-8", errors="ignore") as f:
    contenido = f.read()
    # Utilizar la estructura de datos un conjunto (set) permite eliminar elementos duplicados
    palabras = set(contenido.split())

# Imprimir la cantidad total de esas palabras sin redundancia, denotando el peso del vocabulario del archivo
print(len(palabras))
```
