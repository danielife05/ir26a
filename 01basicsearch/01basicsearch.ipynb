```python
import os
path = r"C:\Users\DELL\OneDrive - Escuela Politécnica Nacional\DANIEL\EPN\SÉPTIMO SEMESTRE\RI\CODIGO\ir26a\data\gutenberg\data"
corpus = os.listdir(path)

textos = {}
for nombre_archivo in corpus:
    file_path = os.path.join(path, nombre_archivo)
    with open(file_path, "r", encoding="utf-8") as file:
        textos[nombre_archivo] = file.read()
        print(f"--- CONTENIDO DE {nombre_archivo} ---")
        print(textos[nombre_archivo][:500] + "\n")
```

```python
import time
query = "Daniel"
start_time = time.time()
resultados = []

for nombre_archivo, contenido_texto in textos.items():
    if query in contenido_texto:
        print(f"El archivo {nombre_archivo} SÍ contiene la palabra '{query}'")
        resultados.append(nombre_archivo)
    else:
        print(f"El archivo {nombre_archivo} NO contiene la palabra '{query}'")

end_time = time.time()
print(f"\nTiempo de búsqueda: {end_time - start_time:.4f} segundos")
print(f"Archivos encontrados: {resultados}")
```

```python
def buscar(query, contenidos):
    resultados = []
    for key in contenidos.keys():
        if query in contenidos[key]:
            resultados.append(key)
    return resultados
```

```python
with open(path + "\\pg84.txt") as f:
    contenido = f.read()
    palabras = contenido.split()

print(len(palabras))
```

```python
start_time = time.time()
for p in palabras:
    buscar(p, textos) # Reemplazado 'contenidos' por 'textos', que es donde guardaste los archivos
end_time = time.time()

# Cambiado el print para que sea un string válido y muestre el cálculo correcto
print(f"tiempo de búsqueda: {(end_time - start_time) * 1000:.2f} milisegundos")
```
